LOADING STEPS: Reading the 3 csv files from bronze storage location and create spark DataFrames

```python
customers = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/ecommerce_data_analysis/bronze/ecommerce_raw/customers.csv")

products = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/ecommerce_data_analysis/bronze/ecommerce_raw/products.csv")

sales = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/ecommerce_data_analysis/bronze/ecommerce_raw/sales.csv")
```

VALIDATION CHECK : Checking how many records where loaded

```python
print("Customers:",customers.count())
print("Products:",products.count())
print("Sales:",sales.count())
```

Creates Delta tabels from the DataFrames we loaded

```python
# Create Bronze Delta tables

customers.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce_data_analysis.bronze.bronze_customers")

products.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce_data_analysis.bronze.bronze_products")

sales.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("ecommerce_data_analysis.bronze.bronze_sales")

print("Bronze tables created successfully!")
```
